```python
def group_strlist(self, strlist: list[str]) -> dict[str, list[str]]:
    """
    Groups strings in strlist by common substrings separated by underscores.
    Returns a dictionary mapping substrings to lists of strings containing them.
    """
    groups = dict()
    groups_idx = []
    count = 0
    for i, s in enumerate(strlist):
        if s not in groups.keys():
            count += 1
            groups[s] = count
        groups_idx.append(groups[s])

    return groups_idx

test = ['aaa', 'aaa', 'bbb', 'aaa', 'bbb', 'ccc', 'ccc']
test = np.array(group_strlist(None, test))
```
